```python
# ASSIGNMENT-2
# QUESTION-1  --LISTS--
roll_no=input("Enter your roll_no:")
L=[]
digit=[]
for i in str(roll_no):
  digit.append(int(i))

L=[i*10 for i in digit]

# (i) Print L
print(L)

# (ii)Add two numbers of your choice to L — one using append() and one using insert() at a specific position.
# Print L after each and explain in a comment what changed.
L.append(90)
print("After append(90):", L)
# append() adds 90 at the end of the list.

L.insert(5, 80)
print("After insert(5, 80):", L)
# insert() adds 80 at index position 5.


# (iii)Remove two elements from L — one using remove() and one using pop(). Print L after each.
L.remove(90)
print("After remove(90):", L)
# remove() removes the first occurrence of the specified value.
element = L.pop()
print("After pop():", L)
print("Element removed using pop():", element)
# pop() removes and returns an element, here the last element.


# (iv)Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.
L.sort()
print("The print is sorted in ascending order:", L)
L.sort(reverse=True)
print("The list is sorted in descending order:",L)


# (v)Use slicing to print the first three elements and the last three elements of L in a single line each.
print("First three elements:",L[:3])
print("Last three elements:",L[-3:])


# (vi)Use list comprehension to create a new list containing only the elements of L that are greater than the
# average of L.
avg=sum(L)/len(L)
greater=[]
for i in L:
  if i>avg:
    greater.append(i)
print("the average is:",avg)
print("elements greater then average:",greater)
```

```python
# QUESTION-2  --TUPLES--
scores=tuple(L[:8])
print("Scores are:",scores)

# (i)Find the highest score and its index in the tuple, and the lowest score and how many times it appears.
# i. Highest score and its index
high_score = max(scores)
high_index = scores.index(high_score)

# Lowest score and its frequency
lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print("Highest score:", high_score)
print("Index of highest score:", high_index)

print("Lowest score:", lowest_score)
print("Number of times lowest score appears:", lowest_count)



# (ii)Reverse the tuple and return the result as a list (tuples themselves cannot be reversed in place —
# explain why in a one-line comment, referring to tuple immutability).
reverse_scores = list(reversed(scores))
print("Reversed tuple as list:", reverse_scores)

# Tuples are immutable, so they cannot be changed/reversed in place.


# (iii)Ask the user to input a score. Print its first occurrence index if present in the tuple, or a not present
# message if not.
search = int(input("Enter a score to search in tuple: "))

if search in scores:
    print("First occurrence index:", scores.index(search))
else:
    print("Score is not present in the tuple.")


# (iv)Attempt to change one element of the tuple directly (e.g., scores[0] = 100). Run it, capture the error
# Python raises, and explain in one line why this happens and how it differs from a list.
scores[0]=100

# Tuple elements cannot be changed because tuples are immutable,
# whereas list elements can be modified.


# (v)Unpack the tuple into a first score, a second score, and all remaining scores using the * operator in a
# single line of unpacking.
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# QUESTION-3  --RANDOM NUMBERS--
import random
random.seed(int(roll_no))

# (i)Generate a list of 100 random numbers between 100 and 900 (inclusive).
random_nums=[random.randint(100,900) for i in range(100)]
print("the 100 random numbers are:")
print(random_nums)



# (ii)Count and print all odd numbers in the list.
odd_numbers=[]
for x in random_nums:
  if x%2!=0:
    odd_numbers.append(x)
print("odd_numbers")
print(odd_numbers)
print("the length of odd numbers is:",len(odd_numbers))


# (iii)Count and print all even numbers in the list.
even_numbers=[]
for x in random_nums:
  if x%2==0:
    even_numbers.append(x)
print("even_numbers")
print(even_numbers)
print("the length of even numbers is:",len(even_numbers))



# (iv)Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
# found using list comprehension.

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in random_nums if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))



# v.Print the number that occurs most frequently in your list, and how many times it occurs.

frequency = {}

for number in random_nums:
    frequency[number] = frequency.get(number, 0) + 1

most_frequent_num= max(frequency, key=frequency.get)
most_frequent_cnt = frequency[most_frequent_num]

print("\nMost frequently occurring number:", most_frequent_num)
print("Number of times it occurs:", most_frequent_cnt)
```

```python
# QUESTION-5  --DICTIONARIES--
#  Consider the following dictionary:
name = input("Enter your name: ")
branch = input("Enter your branch: ")
age = int(input("Enter your age: "))
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": city
}

print("\nOriginal dictionary:")
print(my_dict)



# (i) Rename the key “city” to “location” without changing its value (do not just recreate the dictionary by
# hand — use pop() or dictionary comprehension so the code would work for any dictionary).
my_dict["location"] = my_dict.pop("city")

print("\ni. After renaming city to location:")
print(my_dict)



# (ii)Add a new key “cgpa” with your CGPA as its value.
my_dict["cgpa"] = cgpa

print("\nii. After adding CGPA:")
print(my_dict)


# (iii.) Update the value of “age” by increasing it by 1.
my_dict["age"] = my_dict["age"] + 1

print("\niii. Age after increasing by 1:")
print(my_dict)


# iv. Delete the “branch” key using two different methods in two separate copies of the dictionary: once
# with pop() and once with the del keyword. Explain in one line how pop() differs from del (hint: what
# does each return?).
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Using pop()
removed_branch = dict_pop.pop("branch")

print("\niv. Using pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

# Using del
del dict_del["branch"]

print("Using del:")
print(dict_del)
# pop() removes the key and returns its value,
# while del removes the key but does not return its value.



# v. Iterate over your dictionary using .items() and print each key-value pair as “key → value”.
print("\nv. Key-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)


# (vi.) Before accessing a key called “email” (which doesn't exist in your dictionary), check for its existence
# using the in keyword and print a safe fallback message instead of letting the program crash.
print("\nvi. Checking for email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist in the dictionary.")


# (vii.) Create a second dictionary friend_dict with the same 5 original keys but a friend's (fictional) details.
# Merge it with your own using the {**dict1, **dict2} syntax, and explain in one line which values “win”
# when both dictionaries share a key.
friend_dict = {
    "name": "Khushi verma",
    "roll_no": "1024170065",
    "branch": "CSE",
    "age": 21,
    "city": "Agra"
}

merged_dict = {**my_dict, **friend_dict}

print("\nvii. Friend dictionary:")
print(friend_dict)

print("Merged dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.



# (viii.) Use dictionary comprehension to create a new dictionary containing only the key-value pairs from your
# original dictionary where the value is a string.
string_values_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nviii. Dictionary containing only string values:")
print(string_values_dict)
```

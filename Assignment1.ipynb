```python
# Assignment 5.2 — Add numbers divisible by 7 and 9
n = int(input("Enter value of n: "))

sum = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        sum = sum + i

print("Sum =", sum)
```

```python
# Assignment 5.3 — Add all prime numbers from 1 to n
n = int(input("Enter n: "))

sum = 0

for num in range(2, n + 1):

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        sum = sum + num

print("Sum of prime numbers =", sum)
```

```python
# Assignment 6.1 — Function to add all odd numbers from 1 to n
def add(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum


n = int(input("Enter value of n: "))

result = add(n)

print("Sum of odd numbers =", result)
```

```python
# Assignment 6.2 — Function to add all prime numbers from 1 to n
def prime(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


def add_prime(n):
    sum = 0

    for i in range(1, n + 1):
        if prime(i):
            sum = sum + i

    return sum


n = int(input("Enter n: "))

result = add_prime(n)

print("Sum of prime numbers =", result)
```

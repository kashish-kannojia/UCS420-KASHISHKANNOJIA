```python
# QUESTION6
#  Create a sample dataset (employees.csv) containing information about employees in
# a company.


import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
# WAP to determine the following using dataset in Q6:
# a) Shape (number of rows and columns) of the DataFrame.


print(employees.shape)
```

```python
# b) Summary of the DataFrame that includes the data types and non-null counts for
# each column.

print(employees.info())
```

```python
# c) Generate descriptive statistics.


print(employees.describe())
```

```python
# d) Display the first 5 rows


print(employees.head())
```

```python
# DISPLAY the last 3 rows of the dataset.


print(employees.tail(3))
```

```python
# e) Calculate the following statistics from the dataset:
# i. The average salary of employees.


average = employees["Salary"].mean()

print("Average Salary:", average)
```

```python
# ii. The total bonus paid to all employees.


bonus = employees["Bonus"].sum()

print("Total Bonus:", bonus)
```

```python
# iii. The youngest employee's age.

youngest = employees["Age"].min()

print("Youngest Age:", youngest)
```

```python
# iv. The highest performance rating.

high_rating = employees["Rating"].max()

print("Highest Rating:", high_rating)
```

```python
# f) Sort the DataFrame by the Salary column in descending order.


sorted_employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print(sorted_employees)
```

```python
# g) Add a new column that categorizes employees based on their performance rating:
# i. Excellent for ratings >= 4.5
# ii. Good for ratings >= 4.0 but < 4.5
# iii. Average for ratings < 4.0


def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(
    performance_category
)

print(employees)
```

```python
# h) Identify missing values in the DataFrame.


print(employees.isnull())
```

```python
# i) Rename the Employee_ID column to ID.

employees.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)

print(employees)
```

```python
# j) Find all employees who:
# i. Have more than 5 years of experience.


experienced = employees[
    employees["Years_of_Experience"] > 5
]

print(experienced)
```

```python
# ii. Belong to the IT department.


it_employees = employees[
    employees["Department"] == "IT"
]

print(it_employees)
```

```python
# k) Modify the dataset by adding a new column, Tax, which deducts 10% of the
# Salary.


employees["Tax"] = employees["Salary"] * 0.10

print(employees)
```

```python
# l) Save the modified DataFrame (with added columns) to a new CSV file.

employees.to_csv(
    "employees_modified.csv",
    index=False
)

print("Modified DataFrame saved successfully.")
```
